# Jupyter Notebook Patterns

This notebook demonstrates the key patterns you'll use in Jupyter:
- Loading and exploring data
- Timing and benchmarking
- Visualization with inline plots
- Creating summary tables
- Understanding memory management

**Run each cell with Shift+Enter** to see the output inline.

```python
# Import everything we need
import time
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

print("Libraries imported successfully!")
```

## Cell 1: Load & Explore Data

In a real project, you'd load a CSV or dataset here. We'll create sample data instead.

```python
# Create sample data
data = np.random.randn(1000, 5)
df = pd.DataFrame(data, columns=['feature_' + str(i) for i in range(5)])

print(f"Dataset shape: {df.shape}")
print(f"\nFirst 3 rows:")
df.head(3)
```

```python
# Statistics - in a notebook, this renders as a nice table
print("Data statistics:")
df.describe()
```

## Cell 2: Timing & Benchmarking

Use `%timeit` for micro-benchmarks (runs many times) and `%%time` for one-off operations.

```python
# Time a simple operation
%timeit df.mean()
```

```python
%%time
# Time a longer operation (runs once)
result = df.corr()
print(f"Correlation matrix computed, shape: {result.shape}")
```

## Cell 3: Visualization

Enable inline plotting with `%matplotlib inline` (automatically done in modern Jupyter)

```python
%matplotlib inline

plt.figure(figsize=(12, 4))

# Plot 1: Histogram
plt.subplot(1, 2, 1)
plt.hist(df['feature_0'], bins=30, edgecolor='black', color='skyblue')
plt.title('Distribution of feature_0')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Plot 2: Correlation heatmap
plt.subplot(1, 2, 2)
corr = df.corr()
plt.imshow(corr, cmap='coolwarm', aspect='auto', vmin=-1, vmax=1)
plt.colorbar(label='Correlation')
plt.title('Feature Correlation Matrix')
plt.xticks(range(5), [f'f{i}' for i in range(5)])
plt.yticks(range(5), [f'f{i}' for i in range(5)])

plt.tight_layout()
plt.show()
```

## Cell 4: Summary Tables

DataFrames render as rich HTML tables in notebooks, making comparisons easy.

```python
# Create a model comparison table
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest', 'XGBoost', 'Neural Network'],
    'Accuracy': [0.72, 0.85, 0.88, 0.91],
    'Training Time (s)': [0.1, 2.3, 5.1, 45.6],
    'Parameters': [102, 50_000, 25_000, 1_200_000]
})

# Display the table
results
```

```python
# Use pandas to find the best model
best_idx = results['Accuracy'].idxmax()
best_model = results.loc[best_idx, 'Model']
best_acc = results.loc[best_idx, 'Accuracy']

print(f"Best model: {best_model} with accuracy {best_acc:.2f}")

fastest_idx = results['Training Time (s)'].idxmin()
fastest_model = results.loc[fastest_idx, 'Model']
print(f"Fastest model: {fastest_model}")
```

## Cell 5: Memory Management

Important: Variables persist in the kernel between cells. This is powerful but can leak memory.

```python
# Check memory usage
import sys

small = np.random.randn(1000)
medium = np.random.randn(100_000)
large = np.random.randn(10_000_000)

print("Array sizes:")
for name, arr in [('1K', small), ('100K', medium), ('10M', large)]:
    size_mb = arr.nbytes / 1e6
    print(f"  {name:>4s} elements: {size_mb:>8.2f} MB")

print("\nNote: All these variables are still in memory!")
print("If you're running low on RAM, restart the kernel or use: del large; gc.collect()")
```

## Key Takeaways

1. **Explore in notebooks** — load data, visualize, prototype
2. **Use magic commands** — `%timeit`, `%matplotlib inline`, `!pip install`
3. **Mix code and markdown** — explain your thinking alongside code
4. **Watch for hidden state** — variables persist, can cause bugs if you run out of order
5. **Ship in scripts** — once working, move production code to `.py` files

### Important before sharing a notebook:
- **Kernel > Restart & Run All** — ensures it works top-to-bottom
- Clean up large variables you no longer need
- Document assumptions and dependencies
